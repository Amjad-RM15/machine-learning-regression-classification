```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


file_path = 'CO2 Emissions4.csv'
data = pd.read_csv(file_path)


def plot_categorical(data, column, top_n=None):
    plt.figure(figsize=(10, 6))
    if top_n:
        top_categories = data[column].value_counts().nlargest(top_n).index
        filtered_data = data[data[column].isin(top_categories)]
        sns.countplot(data=filtered_data, y=column, order=top_categories, palette="viridis")
    else:
        sns.countplot(data=data, y=column, order=data[column].value_counts().index, palette="viridis")
    plt.title(f"Distribution of {column}", fontsize=16)
    plt.xlabel("Count", fontsize=12)
    plt.ylabel(column, fontsize=12)
    plt.tight_layout()
    plt.show()

categorical_columns = ['Make', "'Vehicle Class'", 'Transmission', "'Fuel Type'"]

for col in categorical_columns:
    plot_categorical(data, col)

plot_categorical(data, 'Model', top_n=20)  # Display top 20 models
```

1. Model Construction: 
    
o Split the dataset into training (70%) and testing (30%) subsets. 
o Build a Multiple Linear Regression model to predict CO₂ 
Emissions using:  

 Engine Size

 Fuel Consumption (Comb) 

 Cylinders 

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

file_path = 'CO2 Emissions4.csv'
df = pd.read_csv(file_path)

df.rename(columns=lambda x: x.strip().replace("'", "").replace(" ", "_"), inplace=True)

def handle_outliers(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return df[(df[column] >= lower_bound) & (df[column] <= upper_bound)]

columns_to_check = [
    'Engine_Size(L)', 
    'Fuel_Consumption_City_(L/100_km)', 
    'Fuel_Consumption_Hwy_(L/100_km)', 
    'Fuel_Consumption_Comb_(L/100_km)', 
    'CO2_Emissions(g/km)'
]

for column in columns_to_check:
    df = handle_outliers(df, column)

X = df[['Engine_Size(L)', 'Fuel_Consumption_Comb_(L/100_km)', 'Cylinders']]
y = df['CO2_Emissions(g/km)']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

regr = LinearRegression()
regr.fit(X_train, y_train)

print(f"Intercept: {regr.intercept_:.3f}")
print(f"Coefficients: {', '.join([f'{coef:.3f}' for coef in regr.coef_])}")

y_pred = regr.predict(X_test)

mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("\nModel Performance:")
print(f"Mean Squared Error (MSE): {mse:.3f}")
print(f"R-squared (R²): {r2:.3f}")

new_record = pd.DataFrame([[3.5, 8.5, 6]], columns=['Engine_Size(L)', 'Fuel_Consumption_Comb_(L/100_km)', 'Cylinders'])
predicted_co2 = regr.predict(new_record)

print(f"\nPredicted CO2 Emissions for the new record: {predicted_co2[0]:.3f}")
```

2. Feature Selection: 

o Apply feature selection techniques (Forward technique) to 
refine the regression model. 

o Rebuild the model using the selected features.


```python
import statsmodels.api as sm

features = ['Engine_Size(L)', 'Fuel_Consumption_Comb_(L/100_km)', 'Cylinders']
target = 'CO2_Emissions(g/km)'

X = df[features]
y = df[target]

selected_features = []
remaining_features = features.copy()

while remaining_features:
    best_feature = None
    best_score = -float('inf')
    
    for feature in remaining_features:
        current_features = selected_features + [feature]
        X_train_subset = X_train[current_features]
        X_train_subset = sm.add_constant(X_train_subset)  
        
        model = sm.OLS(y_train, X_train_subset).fit()
        score = model.rsquared_adj 
        
        if score > best_score:
            best_feature = feature
            best_score = score
    
    if best_feature:
        selected_features.append(best_feature)
        remaining_features.remove(best_feature)
    else:
        break

print("Selected Features:", selected_features)

X_train_final = sm.add_constant(X_train[selected_features])
model_final = sm.OLS(y_train, X_train_final).fit()

print("\nFinal Model Summary:")
print(model_final.summary())
```

### Tasks: 
1. Preprocess the dataset (handle missing values, encode categorical 
variables).  

I handled the missing data, represented by zeros, by calculating the mean value and imputing it into these entries.

```python
import pandas as pd
import numpy as np
from IPython.display import display 

file_path = 'loan_approval_dataset4.csv'
data = pd.read_csv(file_path)

data.columns = [col.strip().replace("'", "").strip() for col in data.columns]

columns_to_impute = [
    "no_of_dependents",
    "residential_assets_value",
    "commercial_assets_value",
    "bank_asset_value"
]

for col in columns_to_impute:
    mean_value = np.mean(data[col][data[col] != 0])  
    data[col] = data[col].replace(0, mean_value)  

categorical_columns = ["education", "self_employed", "loan_status"]

data_encoded = pd.get_dummies(data, columns=categorical_columns, drop_first=True)

print("Imputed means:")
for col in columns_to_impute:
    print(f"{col}: {np.mean(data[col]):.2f}")

print("\nSample of processed data:")
display(data_encoded.head())  
```

2. Apply various classification

first : DecisionTree

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report

X = data_encoded.drop(columns=['loan_status_\' Rejected\''])  
y = data_encoded['loan_status_\' Rejected\'']  

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

dt_classifier = DecisionTreeClassifier(random_state=42)
dt_classifier.fit(X_train, y_train)

y_pred = dt_classifier.predict(X_test)

print("Decision Tree Classifier:")
print(f"Accuracy: {accuracy_score(y_test, y_pred):.2f}")
print(classification_report(y_test, y_pred))
```

second : LogisticRegression

```python
from sklearn.linear_model import LogisticRegression

lr_classifier = LogisticRegression(max_iter=1000, random_state=42)
lr_classifier.fit(X_train, y_train)

y_pred = lr_classifier.predict(X_test)

print("Logistic Regression:")
print(f"Accuracy: {accuracy_score(y_test, y_pred):.2f}")
print(classification_report(y_test, y_pred))
```

third : Naïve Bayes 

```python
from sklearn.naive_bayes import GaussianNB

nb_classifier = GaussianNB()
nb_classifier.fit(X_train, y_train)

y_pred = nb_classifier.predict(X_test)

print("Naïve Bayes:")
print(f"Accuracy: {accuracy_score(y_test, y_pred):.2f}")
print(classification_report(y_test, y_pred))
```

fourth : RandomForestClassifier (Ensamble)

```python
from sklearn.ensemble import RandomForestClassifier

rf_classifier = RandomForestClassifier(n_estimators=100, random_state=42)
rf_classifier.fit(X_train, y_train)

y_pred = rf_classifier.predict(X_test)

print("Random Forest Classifier:")
print(f"Accuracy: {accuracy_score(y_test, y_pred):.2f}")
print(classification_report(y_test, y_pred))
```

last classifier : SVM with StandardScaler

```python
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, classification_report

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

svm_classifier = SVC(kernel='linear', random_state=42, max_iter=1000)
svm_classifier.fit(X_train_scaled, y_train)

y_pred_svm = svm_classifier.predict(X_test_scaled)

print("Support Vector Machine (with Feature Scaling):")
print(f"Accuracy: {accuracy_score(y_test, y_pred_svm):.2f}")
print(classification_report(y_test, y_pred_svm))
```

```python
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score, precision_score, recall_score, f1_score
import seaborn as sns
import matplotlib.pyplot as plt

classifiers = {
    "Decision Tree": dt_classifier,
    "Logistic Regression": lr_classifier,
    "SVM": svm_classifier,
    "Naive Bayes": nb_classifier,
    "Random Forest": rf_classifier
}

for name, model in classifiers.items():
    y_pred = model.predict(X_test)
    
    cm = confusion_matrix(y_test, y_pred)
    
    acc = accuracy_score(y_test, y_pred)
    prec = precision_score(y_test, y_pred)
    rec = recall_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)
    
    print(f"\n=== Performance Metrics for {name} ===")
    print(f"Accuracy: {acc:.2f}")
    print(f"Precision: {prec:.2f}")
    print(f"Recall: {rec:.2f}")
    print(f"F1-Score: {f1:.2f}")
    print("\nClassification Report:")
    print(classification_report(y_test, y_pred))
    
    plt.figure(figsize=(6, 4))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['Approved', 'Rejected'], yticklabels=['Approved', 'Rejected'])
    plt.title(f"Confusion Matrix for {name}")
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.show()
```

4. Compare hold-out validation (70/30 split) with 10-fold cross
validation. 

```python
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

X = data_encoded.drop(columns=['loan_status_\' Rejected\''])  
y = data_encoded['loan_status_\' Rejected\'']  

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

dt_classifier_holdout = DecisionTreeClassifier(random_state=42)
dt_classifier_holdout.fit(X_train, y_train)

y_pred_holdout = dt_classifier_holdout.predict(X_test)
accuracy_holdout = accuracy_score(y_test, y_pred_holdout)

print("=== Holdout Validation ===")
print(f"Holdout Accuracy: {accuracy_holdout:.3f}")


dt_classifier_cv = DecisionTreeClassifier(random_state=42)
cv_scores = cross_val_score(dt_classifier_cv, X, y, cv=10, scoring='accuracy')

cv_scores_str = " / ".join([f"{score:.3f}" for score in cv_scores])

print("\n=== 10-Fold Cross Validation ===")
print(f"Cross-Validation Accuracy Scores: {cv_scores_str}")
print(f"Mean Accuracy: {cv_scores.mean():.3f}")
print(f"Standard Deviation: {cv_scores.std():.3f}")
```
